```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Load the dataset
df = data

# 1. Data Prep & "Story" Setup
# Identify match boundaries (minute resets to 0)
df['match_id'] = (df['minute'] == 0).cumsum()

# Create a numeric result column for correlation analysis
df['result_numeric'] = df['result'].map({'W': 1, 'D': 0, 'L': -1})

# Calculate ELO difference (Home - Away)
df['elo_diff'] = df['ht_elo'] - df['at_elo']

# Get the final state of each match (the last row per match_id)
match_finals = df.groupby('match_id').last().reset_index()

# Set plot style
sns.set_theme(style="whitegrid", context="talk")

# --- Visual 1: The "Davids vs Goliaths" (ELO Impact) ---
plt.figure(figsize=(10, 6))
# Plotting the distribution of ELO differences for each result
sns.kdeplot(data=match_finals, x='elo_diff', hue='result', 
            palette={'W': '#2ecc71', 'D': '#95a5a6', 'L': '#e74c3c'}, 
            fill=True, alpha=0.3, common_norm=False)
plt.title('Does ELO Dictate Destiny?')
plt.xlabel('Home Team ELO Advantage')
plt.ylabel('Density of Outcomes')
plt.axvline(0, color='black', linestyle='--', alpha=0.3)
plt.savefig('elo_story.png')



```

Before a ball is even kicked, the numbers have already picked a favorite. The first chart visualizes the distribution of match outcomes based on the pre-match ELO difference (Home ELO - Away ELO).

Notice how the green curve swells to the right. When the Home team has a positive ELO advantage ("better" on paper), victory is the dominant outcome.

However, look at the overlap around the 0 mark. Even when teams are evenly matched, the Green curve is slightly higher than the Red curve (Losses). This is the statistical footprint of Home Field Advantage.

There is a long tail of red on the right side. These are the "upsets" games where the better Home team still lost, proving that while ELO predicts probability, it doesn't guarantee destiny.

```python
# --- Visual 3: The "Killer Instinct" (Efficiency vs Volume) ---
# Scatter plot of Shots (Volume) vs Big Chances (Quality) colored by Result
# We use the final match stats
plt.figure(figsize=(10, 8))
sns.scatterplot(data=match_finals, x='shot', y='big_chance', hue='result', 
                palette={'W': '#2ecc71', 'D': '#95a5a6', 'L': '#e74c3c'}, 
                style='result', alpha=0.7, s=100)
plt.title('Quality vs. Quantity: Shots vs. Big Chances')
plt.xlabel('Shot Differential (Home - Away)')
plt.ylabel('Big Chance Differential')
plt.axhline(0, color='black', linestyle='--', alpha=0.3)
plt.axvline(0, color='black', linestyle='--', alpha=0.3)
plt.savefig('efficiency.png')
```

Possession and passing are meaningless without the final ball. The "Quality vs. Quantity" scatter plot pits shot volume against big_chance creation.

We might expect some teams to win with few shots (defensive masterclasses), but the Green dots are overwhelmingly concentrated in the top-right quadrant.

To win, we generally need both positive shot volume (X-axis > 0) and, more importantly, positive "Big Chance" creation (Y-axis > 0).

Notice the few Red dots mixed in with the Green in the top-right? These are the tragedies of football, in which Home teams that out-shot and out-created their opponents but still lost (likely due to poor finishing or bad luck).

```python


# Preprocessing
df['match_id'] = (df['minute'] == 0).cumsum()
match_finals = df.groupby('match_id').last().reset_index()
df['result_numeric'] = df['result'].map({'W': 1, 'D': 0, 'L': -1})

# Set style
sns.set_theme(style="whitegrid", context="talk")

# --- Insight 1: The "Desperation" (Crosses) ---
# Hypothesis: Losing teams spam crosses in the second half.
# We look at the 'cross' differential over time.
cross_time = df.groupby(['minute', 'result'])['cross'].mean().reset_index()

plt.figure(figsize=(12, 6))
sns.lineplot(data=cross_time, x='minute', y='cross', hue='result', 
             palette={'W': '#2ecc71', 'D': '#95a5a6', 'L': '#e74c3c'}, linewidth=3)
plt.title('Crossing Volume Over Time')
plt.xlabel('Minute')
plt.ylabel('Cross Differential (Home - Away)')
plt.legend(title='Home Result')
plt.savefig('story_crosses.png')
```

At the first half, the teams that eventually Win (Green) are actually crossing the ball more than the teams that lose, with +2.88 differential vs. +0.28.

The Narrative: Winning teams use crosses as a surgical tool early in the game to break the deadlock and take the lead. They are crossing with purpose.

Once the second half begins, the lines diverge dramatically.

1, The Winners: The line goes flat. They stop crossing while winning. They switch to "Control" mode, which include keeping the ball on the ground, killing the clock, and refusing to give away possession cheaply.

2, The Losers: The line rockets upward, jumping from near zero to +3.96. This is the sound of a team hitting the panic button. They are chasing the game, the clock is ticking, and their tactical discipline collapses. They stop trying to craft goals and start blindly flinging the ball into the box.

```python
# --- Insight 4: The "Territorial Illusion" (Final 3rd vs Goals) ---
# Does dominating the Final 3rd actually guarantee goals?
# Calculate goal difference
match_finals['goal_diff'] = match_finals['ht_goal'] - match_finals['at_goal']

plt.figure(figsize=(10, 8))
sns.scatterplot(data=match_finals, x='final_3rd_pass', y='goal_diff', hue='result',
                palette={'W': '#2ecc71', 'D': '#95a5a6', 'L': '#e74c3c'}, alpha=0.6)
plt.title('Final 3rd Passes vs. Actual Goals')
plt.xlabel('Final 3rd Pass Differential')
plt.ylabel('Goal Difference')
plt.axvline(0, color='black', linestyle='--', alpha=0.3)
plt.axhline(0, color='black', linestyle='--', alpha=0.3)
plt.savefig('story_territory.png')
```

Can you dominate the pitch and still lose? Yes, and it's heartbreaking.

The scatter plot compares final_3rd_pass (X-axis) with goal_diff (Y-axis).

Look at the Red dots in the bottom right. These are teams that had a massive positive differential in Final 3rd passes ( +50 to +100) but still had a negative goal difference.

In essence, they "camped" in the opponent's half but couldn't score.

The Lesson: Territory is vanity. You can own the final third and still lose if you lack the Big Chances (1 on 1 situation) and scoring ability.

```python



# --- Insight 5: The "Architect" (Key Passes vs Big Chances) ---
# Connecting creativity to quality.
plt.figure(figsize=(10, 8))
sns.regplot(data=match_finals, x='key_pass', y='big_chance', 
            scatter_kws={'alpha':0.3, 'color': 'purple'}, line_kws={'color': 'orange'})
plt.title('Key Passes converting to Big Chances')
plt.xlabel('Key Pass Differential')
plt.ylabel('Big Chance Differential')
plt.savefig('story_creativity.png')
```

So, what should happen inside that 3rd corner of the field? We need to create more chances.

The regression plot shows the link between key_pass and big_chance.

The correlation is incredibly strong and linear.

This suggests that big_chance (1v1 situations) isn't random. It is almost directly manufactured by key_pass volume.

The Lesson: You don't get lucky scoring chances. You build them, pass by pass. If you want more Big Chances, you don't need "better finishers", you need more Key Passes.
